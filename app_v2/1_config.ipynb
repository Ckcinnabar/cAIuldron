```python
import os
from pathlib import Path
from dotenv import load_dotenv

# Load .env from app_v2/ directory
_notebook_dir = Path(os.getcwd())
# Support running from project root OR from app_v2/
if (_notebook_dir / 'app_v2' / '.env').exists():
    load_dotenv(_notebook_dir / 'app_v2' / '.env')
elif (_notebook_dir / '.env').exists():
    load_dotenv(_notebook_dir / '.env')

print('Environment loaded')
```

```python
# ── Project Paths ─────────────────────────────────────────────────────────
PROJECT_ROOT = Path(os.getcwd())
# Adjust if cwd is app_v2/ instead of project root
if (PROJECT_ROOT / 'data').exists():
    pass  # already at project root
elif (PROJECT_ROOT.parent / 'data').exists():
    PROJECT_ROOT = PROJECT_ROOT.parent

DATA_DIR        = PROJECT_ROOT / 'data'
RECIPES_JSON    = DATA_DIR / 'processed' / 'recipes' / 'full_recipes.json'
NUTRITION_JSON  = DATA_DIR / 'nutrition_lookup_full.json'
APP_V2_DIR      = PROJECT_ROOT / 'app_v2'
CHROMA_DIR      = APP_V2_DIR / '.chroma_db'

print(f'Project root : {PROJECT_ROOT}')
print(f'Recipes JSON : {RECIPES_JSON.exists()} → {RECIPES_JSON}')
print(f'Nutrition DB : {NUTRITION_JSON.exists()} → {NUTRITION_JSON}')
print(f'ChromaDB dir : {CHROMA_DIR}')
```

```python
# ── API Keys ──────────────────────────────────────────────────────────────
GROQ_API_KEY        = os.environ.get('GROQ_API_KEY', '')
TAVILY_API_KEY      = os.environ.get('TAVILY_API_KEY', '')
HF_TOKEN            = os.environ.get('HF_TOKEN', '')
LANGCHAIN_API_KEY   = os.environ.get('LANGCHAIN_API_KEY', '')

# Set LangSmith tracing env vars (LangGraph reads these automatically)
os.environ['LANGCHAIN_TRACING_V2']  = os.environ.get('LANGCHAIN_TRACING_V2', 'false')
os.environ['LANGCHAIN_PROJECT']     = os.environ.get('LANGCHAIN_PROJECT', 'cAIuldron-v2')

# Validate keys
missing = [k for k, v in {
    'GROQ_API_KEY': GROQ_API_KEY,
    'TAVILY_API_KEY': TAVILY_API_KEY,
    'HF_TOKEN': HF_TOKEN,
}.items() if not v]

if missing:
    print(f'⚠️  Missing API keys: {missing}')
    print('   Copy app_v2/.env.example → app_v2/.env and fill in your keys')
else:
    print('✅ All API keys loaded')
print(f'   LangSmith tracing: {os.environ["LANGCHAIN_TRACING_V2"]}')
```
